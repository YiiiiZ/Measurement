```python
from collections import deque
class Solution:
    def maxSlidingWindow(self, nums, k):
        # Deque holds indices of elements in decreasing value.
        # The front of the deque is always the max of the current window.
        dq = deque()
        out = []

        for i, val in enumerate(nums):
            print(dq)
            # Remove indices whose values are smaller than the new value
            # because they can never be a future maximum.
            while dq and nums[dq[-1]] <= val:
                dq.pop()

            # Add current index.
            dq.append(i)

            # Remove the index at the front if it is outside the window.
            if dq[0] <= i - k:
                dq.popleft()

            # Once we have the first full window, record the max each step.
            if i >= k - 1:
                out.append(nums[dq[0]])

        return out
    
s = Solution()
print(s.maxSlidingWindow([1,3,-1,-3,5,3,6,7], 3))
```
